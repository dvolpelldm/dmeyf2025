# Genera distintos cortes del prediccion.txt

```r
# limpio la memoria
Sys.time()
rm(list=ls(all.names=TRUE)) # remove all objects
gc(full=TRUE, verbose=FALSE) # garbage collection
```

```r
PARAM <- list()
PARAM$experimento <- "apo-004"
PARAM$semilla_primigenia <- 102191
```

```r
setwd("/content/buckets/b1/exp")
experimento_folder <- PARAM$experimento
dir.create(experimento_folder, showWarnings=FALSE)
setwd( paste0("/content/buckets/b1/exp/", experimento_folder ))
```

```r
library(data.table)

# 1) Leo la predicción desde disco
tb_prediccion <- fread("prediccion.txt", sep = "\t")

# 2) Ordeno por probabilidad descendente
setorder(tb_prediccion, -prob)

# 3) Me aseguro de tener la carpeta 'kaggle'
dir.create("kaggle", showWarnings = FALSE)

# 4) Valores de envíos que quiero probar
envios_vec <- c(9500, 10000, 10500, 11000, 11500)

# 5) Genero un CSV por cada cantidad de envíos
for (envios in envios_vec) {

  # arranco de cero cada vez
  tb_prediccion[, Predicted := 0L]
  tb_prediccion[1:envios, Predicted := 1L]

  archivo_kaggle <- paste0("./kaggle/KA", PARAM$experimento, "_", envios, ".csv")

  fwrite(
    tb_prediccion[, .(numero_de_cliente, Predicted)],
    file = archivo_kaggle,
    sep = ","
  )

  cat("Generado:", archivo_kaggle, "\n")
}
```
